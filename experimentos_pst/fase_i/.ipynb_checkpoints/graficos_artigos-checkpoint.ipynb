```python
import pandas as pd
import numpy as np
```

```python
def bootstrap_ci(x, stat=np.mean, n_boot=5000, alpha=0.05, seed=42):
    rng = np.random.default_rng(seed)
    x = np.asarray(x, dtype=float)
    x = x[~np.isnan(x)]
    if len(x) == 0:
        return np.nan, np.nan
    boots = rng.choice(x, size=(n_boot, len(x)), replace=True)
    stats = np.apply_along_axis(stat, 1, boots)
    lo = np.quantile(stats, alpha / 2)
    hi = np.quantile(stats, 1 - alpha / 2)
    return float(lo), float(hi)

def select_sizes(df, must_keep=(8, 1024, 2048), start=2048, step=2048):
    sizes = np.sort(df["size_bits"].unique())
    keep = set()

    # mantém âncoras se existirem
    for s in must_keep:
        if s in sizes:
            keep.add(int(s))

    # mantém grade a partir de start
    for s in sizes:
        if s >= start and (s - start) % step == 0:
            keep.add(int(s))

    return sorted(keep)

def summarize_for_pgf(df: pd.DataFrame, sizes_keep=None, add_bootstrap_ci=True) -> pd.DataFrame:
    df = df.copy()
    df["size_bits"] = df["size_bits"].astype(int)
    df["total_time_s"] = pd.to_numeric(df["total_time_s"], errors="coerce")
    df["throughput_bps"] = pd.to_numeric(df["throughput_bps"], errors="coerce")

    ok = df[df["error_code"] == "OK"].copy()

    # IMPORTANTE: groupby por string (chave escalar, não tupla)
    g = ok.groupby("size_bits", dropna=False)

    # Iguais ao seu summarize()
    summary = g.agg(
        n=("total_time_s", "size"),
        time_mean_s=("total_time_s", "mean"),
        time_std_s=("total_time_s", "std"),
        thr_mean_bps=("throughput_bps", "mean"),
        thr_std_bps=("throughput_bps", "std"),
    ).reset_index()

    qdf = g["total_time_s"].quantile([0.50, 0.95, 0.99]).unstack()
    qdf = qdf.rename(columns={0.50: "time_p50_s", 0.95: "time_p95_s", 0.99: "time_p99_s"}).reset_index()
    summary = summary.merge(qdf, on="size_bits", how="left")

    # Taxa de erro (considera tudo)
    err = (
        df.assign(is_ok=(df["error_code"] == "OK"))
          .groupby("size_bits")["is_ok"]
          .agg(n_total="size", n_ok="sum")
          .reset_index()
    )
    err["error_rate_pct"] = (1.0 - (err["n_ok"] / err["n_total"])) * 100.0
    summary = summary.merge(err[["size_bits", "n_total", "error_rate_pct"]], on="size_bits", how="left")

    # IC bootstrap (opcional)
    if add_bootstrap_ci:
        ci_rows = []
        for size_bits, gg in g:
            t = gg["total_time_s"].to_numpy(dtype=float)
            thr = gg["throughput_bps"].to_numpy(dtype=float)

            t_lo, t_hi = bootstrap_ci(t, np.mean)
            thr_lo, thr_hi = bootstrap_ci(thr, np.mean)

            ci_rows.append({
                "size_bits": int(size_bits),
                "time_ci95_lo_s": t_lo,
                "time_ci95_hi_s": t_hi,
                "thr_ci95_lo_bps": thr_lo,
                "thr_ci95_hi_bps": thr_hi,
            })

        ci_df = pd.DataFrame(ci_rows)
        summary = summary.merge(ci_df, on="size_bits", how="left")

    summary = summary.sort_values("size_bits").reset_index(drop=True)

    # Filtra no final
    if sizes_keep is not None:
        sizes_keep = set(int(x) for x in sizes_keep)
        summary = summary[summary["size_bits"].isin(sizes_keep)].reset_index(drop=True)

    return summary

def export_pgf_csvs(
    summary: pd.DataFrame,
    time_csv_path: str = "time_vs_size_pgf.csv",
    thr_csv_path: str = "throughput_vs_size_pgf.csv",
):
    """
    Exporta CSVs prontos para PGFPlots.
    - time_vs_size_pgf.csv: média + IC 95%
    - throughput_vs_size_pgf.csv: throughput médio
    """

    # Tempo com IC
    time_df = summary[
        ["size_bits", "time_mean_s", "time_ci95_lo_s", "time_ci95_hi_s"]
    ].copy()

    time_df.to_csv(time_csv_path, index=False)

    # Throughput médio
    thr_df = summary[
        ["size_bits", "thr_mean_bps"]
    ].copy()

    thr_df.to_csv(thr_csv_path, index=False)

    return time_csv_path, thr_csv_path
```

```python
df = pd.read_csv("sbrc_experiment_official.csv")

# garante tipos numéricos relevantes
df["size_bits"] = df["size_bits"].astype(int)
df["total_time_s"] = df["total_time_s"].astype(float)
df["throughput_bps"] = df["throughput_bps"].astype(float)

sizes_keep = select_sizes(df, must_keep=(8, 1024, 2048), start=2048, step=1024)
summary = summarize_for_pgf(df, sizes_keep=sizes_keep)

time_path, thr_path = export_pgf_csvs(summary)
print("sizes_keep =", sizes_keep)
print("Wrote:", time_path, thr_path)
print(summary)
```
