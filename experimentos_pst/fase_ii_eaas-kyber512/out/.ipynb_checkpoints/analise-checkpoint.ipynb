```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
csv_path = Path("results.csv")
df = pd.read_csv(csv_path)

print("linhas:", len(df))
df.head()
```

```python
# Colunas esperadas (ajusta conforme seu CSV)
required = ["mode", "error_code"]
missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"CSV não tem colunas mínimas {missing}. Colunas atuais: {list(df.columns)}")

# Tenta converter tempos (se existirem)
for c in ["t_seed_ms", "t_kem_ms", "t_total_ms"]:
    if c in df.columns:
        df[c] = pd.to_numeric(df[c], errors="coerce")

# Converte bytes_len se existir
if "entropy_bytes_len" in df.columns:
    df["entropy_bytes_len"] = pd.to_numeric(df["entropy_bytes_len"], errors="coerce")

# Normaliza mode
df["mode"] = df["mode"].astype(str).str.strip().str.lower()

# Define sucesso
df["ok"] = df["error_code"].astype(str).str.upper().eq("OK")

df["mode"].value_counts(dropna=False), df["ok"].value_counts(dropna=False)
```

```python
EXPECTED_BYTES = 256

if "entropy_bytes_len" in df.columns:
    for m in ["eaas", "direct"]:
        if (df["mode"] == m).any():
            bad = df[(df["mode"] == m) & df["ok"] & (df["entropy_bytes_len"] != EXPECTED_BYTES)]
            print(f"[{m}] OK com bytes != {EXPECTED_BYTES}: {len(bad)}")
            if len(bad):
                display(bad[["mode","entropy_bytes_len","error_code","t_seed_ms","t_total_ms"]].head(10))
else:
    print("Coluna entropy_bytes_len não existe; pulando sanity check de bytes.")
```

```python
time_cols = [c for c in ["t_seed_ms", "t_kem_ms", "t_total_ms"] if c in df.columns]
if not time_cols:
    raise ValueError("CSV não tem colunas de tempo (t_seed_ms/t_kem_ms/t_total_ms).")
print("Tempos disponíveis:", time_cols)
```

```python
def summarize_mode(g: pd.DataFrame) -> dict:
    out = {}
    out["n"] = len(g)
    out["n_ok"] = int(g["ok"].sum())
    out["error_pct"] = 100.0 * (1 - out["n_ok"]/out["n"]) if out["n"] else np.nan

    gok = g[g["ok"]].copy()
    for c in ["t_seed_ms", "t_kem_ms", "t_total_ms"]:
        if c in g.columns:
            x = gok[c].dropna().to_numpy()
            out[f"{c}_mean"] = float(np.mean(x)) if len(x) else np.nan
            out[f"{c}_median"] = float(np.median(x)) if len(x) else np.nan
            out[f"{c}_p50"] = float(np.percentile(x, 50)) if len(x) else np.nan
            out[f"{c}_p95"] = float(np.percentile(x, 95)) if len(x) else np.nan
            out[f"{c}_p99"] = float(np.percentile(x, 99)) if len(x) else np.nan
            out[f"{c}_std"] = float(np.std(x, ddof=1)) if len(x) > 1 else np.nan
    return out

summary = (
    df.groupby("mode", dropna=False)
      .apply(summarize_mode)
      .apply(pd.Series)
      .sort_values("t_total_ms_mean" if "t_total_ms_mean" in df.columns else "error_pct")
)

summary
```

```python
SIZE_BITS = 2048

df2 = df.copy()
# throughput efetivo do ciclo completo (operação inteira)
df2["throughput_total_bps"] = np.where(
    df2["t_total_ms"].notna() & (df2["t_total_ms"] > 0),
    SIZE_BITS / (df2["t_total_ms"] / 1000.0),
    np.nan
)

# throughput apenas da etapa de seed (útil para EaaS/direct)
if "t_seed_ms" in df2.columns:
    df2["throughput_seed_bps"] = np.where(
        df2["t_seed_ms"].notna() & (df2["t_seed_ms"] > 0),
        SIZE_BITS / (df2["t_seed_ms"] / 1000.0),
        np.nan
    )

thr = (
    df2[df2["ok"]]
    .groupby("mode")[["throughput_total_bps"] + (["throughput_seed_bps"] if "throughput_seed_bps" in df2.columns else [])]
    .agg(["mean","median","std"])
)

thr
```

## Boxplot da latência total por modo

```python
ok = df2[df2["ok"] & df2["t_total_ms"].notna()].copy()

modes_order = [m for m in ["baseline","eaas","direct"] if m in ok["mode"].unique()]
data = [ok.loc[ok["mode"] == m, "t_total_ms"].to_numpy() for m in modes_order]

plt.figure()
plt.boxplot(data, labels=modes_order, showfliers=False)
plt.ylabel("t_total_ms")
plt.grid(True, axis="y", linestyle="--", linewidth=0.5)
plt.title("Latência total por modo (apenas execuções OK)")
plt.savefig(f"eaas_latencia_total.pdf", bbox_inches="tight")
plt.show()
```

```python
lat = summary[[c for c in summary.columns if c.startswith("t_total_ms_p")]].copy()
lat = lat.loc[modes_order]

plt.figure()
x = np.arange(len(lat.index))
p95 = lat["t_total_ms_p95"].to_numpy()
p99 = lat["t_total_ms_p99"].to_numpy()

plt.bar(x - 0.15, p95, width=0.3, label="p95")
plt.bar(x + 0.15, p99, width=0.3, label="p99")
plt.xticks(x, lat.index)
plt.ylabel("t_total_ms (ms)")
plt.grid(True, axis="y", linestyle="--", linewidth=0.5)
plt.legend()
plt.title("Tail latency por modo (apenas OK)")
plt.show()
```

```python
err = summary["error_pct"].loc[modes_order]

plt.figure()
plt.bar(err.index, err.to_numpy())
plt.ylabel("Taxa de erro (%)")
plt.grid(True, axis="y", linestyle="--", linewidth=0.5)
plt.title("Taxa de erro por modo")
plt.show()
```

```python
if "t_seed_ms" in df2.columns and "t_kem_ms" in df2.columns:
    comp = (
        df2[df2["ok"] & df2["mode"].isin(["eaas","direct"])]
        .groupby("mode")[["t_seed_ms","t_kem_ms"]]
        .median()
        .loc[[m for m in ["eaas","direct"] if m in df2["mode"].unique()]]
    )

    plt.figure()
    x = np.arange(len(comp.index))
    plt.bar(x, comp["t_seed_ms"], label="t_seed_ms (mediana)")
    plt.bar(x, comp["t_kem_ms"], bottom=comp["t_seed_ms"], label="t_kem_ms (mediana)")
    plt.xticks(x, comp.index)
    plt.ylabel("Tempo (ms)")
    plt.grid(True, axis="y", linestyle="--", linewidth=0.5)
    plt.legend()
    plt.title("Decomposição do tempo: aquisição de entropia vs KEM")
    plt.savefig(f"eaas_seed_kem.pdf", bbox_inches="tight")
    plt.show()
```

```python
def bootstrap_diff_median(a, b, B=5000, seed=0):
    rng = np.random.default_rng(seed)
    a = np.asarray(a); b = np.asarray(b)
    a = a[np.isfinite(a)]; b = b[np.isfinite(b)]
    if len(a) == 0 or len(b) == 0:
        return np.nan, (np.nan, np.nan)

    diffs = []
    for _ in range(B):
        aa = rng.choice(a, size=len(a), replace=True)
        bb = rng.choice(b, size=len(b), replace=True)
        diffs.append(np.median(aa) - np.median(bb))
    diffs = np.sort(diffs)
    return float(np.median(diffs)), (float(diffs[int(0.025*B)]), float(diffs[int(0.975*B)]))

ok_lat = ok[["mode","t_total_ms"]].dropna()

pairs = [("eaas","baseline"), ("direct","baseline"), ("direct","eaas")]
for a,b in pairs:
    if a in ok_lat["mode"].unique() and b in ok_lat["mode"].unique():
        A = ok_lat.loc[ok_lat["mode"]==a, "t_total_ms"].to_numpy()
        B = ok_lat.loc[ok_lat["mode"]==b, "t_total_ms"].to_numpy()
        med, ci = bootstrap_diff_median(A, B)
        print(f"Δmediana t_total_ms ({a}-{b}) = {med:.2f} ms | IC95% [{ci[0]:.2f}, {ci[1]:.2f}]")
```

```python
cols = []
for c in ["t_total_ms_mean","t_total_ms_p95","t_total_ms_std","t_kem_ms_mean","t_seed_ms_mean","error_pct","n_ok","n"]:
    if c in summary.columns:
        cols.append(c)

table = summary.loc[modes_order, cols].copy()

# arredondamento
for c in table.columns:
    if c.startswith("t_") or c.endswith("_pct"):
        table[c] = table[c].astype(float).round(3)

table
```

```python
latex = table.to_latex(
    index=True,
    escape=True,
    column_format="l" + "r"*len(table.columns),
    caption="Resumo de desempenho por modo (apenas execuções OK para métricas de tempo).",
    label="tab:perf-modes"
)
print(latex)
```
